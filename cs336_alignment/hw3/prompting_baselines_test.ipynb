```python
with open("/root/autodl-tmp/gaoguoji-cs336-assignment5-alignment/data/gsm8k/test.jsonl", "r") as f:
    valid_data = [json.loads(line) for line in f]
```

```python
question_only_promt_template = """{question} Please put your final answer within \\boxed{{}}."""
zero_shot_prompt_template = """A conversation between User and Assistant. The User asks a question, and the Assistant solves it. The Assistant first thinks about the reasoning process in the mind and then provides the User with the answer. The reasoning process is enclosed within <think> </think> and answer is enclosed within <answer> </answer> tags, respectively, i.e., <think> reasoning process here </think> <answer> answer here </answer>.
User: {question}
Assistant: <think>""" 
few_shot_promt_template = """A conversation between User and Assistant. The User asks a question, and the Assistant solves it. The Assistant first thinks about the reasoning process in the mind and then provides the User with the answer. The reasoning process is enclosed within <think> </think> and answer is enclosed within <answer> </answer> tags, respectively, i.e., <think> reasoning process here </think> <answer> answer here </answer>.
User: There are 15 trees in the grove. Grove workers will plant trees in the grove today. After they are done, there will be 21 trees. How many trees did the grove workers plant today?
Assistant: <think> There are 15 trees originally. Then there were 21 trees after some more were planted. So there must have been 21 - 15 = 6. So the answer is 6. </think> <answer> 6 </answer>
User: If there are 3 cars in the parking lot and 2 more cars arrive, how many cars are in the parking lot?
Assistant: <think> There are originally 3 cars. 2 more cars arrive. 3 + 2 = 5. So the answer is 5. </think> <answer> 5 </answer>
User: Leah had 32 chocolates and her sister had 42. If they ate 35, how many pieces do they have left in total?
Assistant: <think> Originally, Leah had 32 chocolates. Her sister had 42. So in total they had 32 + 42 = 74. After eating 35, they had 74 - 35 = 39. So the answer is 39. </think> <answer> 39 </answer>
User: {question}
Assistant: <think>"""
question_only_promt = [question_only_promt_template.format(question=item['question']) for item in valid_data]
zero_shot_prompts = [zero_shot_prompt_template.format(question=item['question']) for item in valid_data] 
few_shot_promts = [few_shot_promt_template.format(question=item['question']) for item in valid_data] 

#llm.generate(valid_data[0]['question'], sampling_params=sampling_params)
```

```python
print(question_only_promt[0])
print(zero_shot_prompts[1])
print(few_shot_promts[2])
```

```python
query_only_sampling_params = SamplingParams(
    temperature=1.0,
    top_p=1.0,
    max_tokens=512
)
zero_few_shot_sampling_params = SamplingParams(
    temperature=1.0,
    top_p=1.0,
    max_tokens=512,
    stop=["</answer>"], # 遇到结束标签即停止
    include_stop_str_in_output=True
) 
```

```python
question_only_responses = llm.generate(question_only_promt, sampling_params=query_only_sampling_params)
question_only_responses[0].outputs[0].text
```

```python
zero_shot_output = llm.generate(zero_shot_prompts, sampling_params=zero_few_shot_sampling_params)
# output = llm.generate(few_shot_promts, sampling_params=zero_few_shot_sampling_params)
```

```python
zero_shot_output[0].outputs[0].text
```

```python
few_shot_output = llm.generate(few_shot_promts, sampling_params=zero_few_shot_sampling_params)
```

```python
few_shot_output[0].outputs[0].text
```

```python
from cs336_alignment.drgrpo_grader import question_only_reward_fn, r1_zero_reward_fn
```

```python
ground_truth = [item['answer'].split("####")[-1].strip() for item in valid_data]
```

```python
few_shot_reward = []
for response, gt in zip(few_shot_output, ground_truth):
    reward = r1_zero_reward_fn(response.outputs[0].text, gt)
    few_shot_reward.append(reward)
```
